```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)#من اجل عرض الارقام بصورة مضبوطة
import sympy as sp
from roboticstoolbox import DHRobot, RevoluteDH


links = [# تعريف الوصلات
    RevoluteDH(a=0.047, alpha=np.pi/2, d=0.05,qlim=[-np.pi/2, np.pi/2]),
    RevoluteDH(a=0.12, alpha=0, d=0,offset=np.pi/2,qlim=[-1.2217,1.2217]),#-70 to 70 degree
    RevoluteDH(a=0.044, alpha=np.pi/2, d=0,qlim=[-1.2217,1.2217]),#-70 to 70 degree
    RevoluteDH(a=0, alpha=np.pi/2, d=0.15,qlim=[-np.pi/2, np.pi/2]),
    RevoluteDH(a=0.038, alpha=0, d=0,offset=np.pi/2,qlim=[-1.2217,1.2217]),#-70 to 70 degree
]
My_robot = DHRobot(links, name="MyRobot")# تعريف الروبوتT\

print(My_robot)# طباعة مواصفات الروبوت
q=[0,0,0,0,0]# تعريف زاوية كل وصلة
T=My_robot.fkine(q)# حساب المصفوفة التحويلية للروبوت
print(T)


#T_tadget=np.array([[0.9209,0.3384,-0.1934,30.03],الهدف
    #[0.0237,0.4468,0.8943,14.58],
    #[0.3890,-0.8282,0.4034,7.829],
     #    [0,0,0,1]])





```

```python
# inverse kinematics

IK_sol1=My_robot.ikine_LM(T, q0=[0,0,0,0,0]) #قيمة ابتدائية للحل الاول
IK_sol2=My_robot.ikine_LM(T, q0 = [(qlim[0]+qlim[1])/2 for l in My_robot.links for qlim in [l.qlim]]
)#قيمة ابتدائية للحل الثاني من منتصف مجال الحركة لكل وصلة
Angle_solution1=IK_sol1.q# الحل الاول بالراديان
Angle_solution2=IK_sol2.q# الحل الثاني بالراديان

for i, link in enumerate(My_robot.links):# ضبط الحلول ضمن حدود الحركة للوصلات
    Angle_solution1[i] = np.clip(q[i], link.qlim[0], link.qlim[1])
    Angle_solution2[i] = np.clip(q[i], link.qlim[0], link.qlim[1])
    
Angle_solution_deg1=np.degrees(Angle_solution1)# تحويل الحل الاول الى درجات
Angle_solution_deg2=np.degrees(Angle_solution2)# تحويل الحل الثاني الى درجات
print(f"Angle_solution1={Angle_solution_deg1}")
print(f"Angle_solution2={Angle_solution_deg2}")




#My_robot.plot(q)

```
